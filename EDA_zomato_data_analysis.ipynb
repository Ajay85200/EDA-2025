# Remove "New","-" and "/5" from Rate column

```python
def handlerate(value):
    if (values =='NEW' or value=="-"):
        return np.nan
    else:
        value =str(value).split('/')
        value = value[0]
        return float(value)

df['rate'] = df['rate'].apply(handlerate)
df['rate'].head()
```
